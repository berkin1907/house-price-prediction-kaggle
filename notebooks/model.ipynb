```python
import pandas as pd
import numpy as np
from sklearn.linear_model import Ridge, Lasso
from sklearn.model_selection import cross_val_score, KFold
```

```python
X_train = pd.read_csv('../data/processed/X_train_processed.csv', index_col=0)
y_train = pd.read_csv('../data/processed/y_train_processed.csv', index_col=0)
X_test  = pd.read_csv('../data/processed/X_test_processed.csv', index_col=0)
```

```python
model = Ridge(random_state=42)
scores = cross_val_score(
    model,
    X_train,
    y_train,
    cv=KFold(5, shuffle=True, random_state=42),
    scoring='neg_root_mean_squared_error'
)
print("5-fold RMSEs:", -scores)
print("Mean RMSE:", -scores.mean())
```

```python
log_target = True 

## From Grid Search CV results
model = Lasso(alpha=0.002395026619987486, random_state=42)
model.fit(X_train, y_train)

preds = model.predict(X_test)
if log_target:
    preds = np.expm1(preds)

submission = pd.DataFrame({
    "Id": X_test.index,
    "SalePrice": preds
})
submission.to_csv("../results/submissions/submission_lasso.csv", index=False)
```
